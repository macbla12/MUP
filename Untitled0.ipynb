```python
import PyMPDATA as pym
import numpy as np
import matplotlib.pyplot as plt
print(PyMPDATA.__file__)
```

```python
x = np.linspace(-2, 8, 41)
y = np.exp(-x**2)-1

plt.figure(figsize=(8, 5))
plt.plot(x, y, color='red', linewidth=2.5, label=r'$y = e^{-x^2}$')

plt.title('Wykres funkcji Gaussa', fontsize=14, fontweight='bold', pad=15)
plt.xlabel('x', fontsize=12)
plt.ylabel('y', fontsize=12)
plt.grid(True, linestyle='--', alpha=0.6)
plt.legend(fontsize=12)

plt.tight_layout()
plt.show()
```

```python
from PyMPDATA import ScalarField, VectorField, Options, Stepper, Solver, boundary_conditions

common = {
    'halo': (options := Options(nonoscillatory=True)).n_halo,
    'boundary_conditions': (boundary_conditions.Periodic(),)
}
advector = VectorField(**common, data=(np.full(x.size + 1, 1/4),))
adscalar = ScalarField(**common, data=y)
solver = Solver(advectee=adscalar, advector=advector, stepper=Stepper(n_dims=1, options=options))
solver.advance(n_steps=100)
```

```python
plt.plot()
plt.step(x,solver.advectee.get(),where='mid')
plt.axhline(-1, color='red')
plt.show()
```
